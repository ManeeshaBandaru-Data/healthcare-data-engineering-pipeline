```python
from pyspark.sql import Row
from datetime import date, timedelta

# Create sample patient data
patients = []

cities = ['Hyderabad', 'Bangalore', 'Chennai', 'Mumbai', 'Pune']
genders = ['Male', 'Female']
insurance_types = ['Private', 'Government', 'Employer']

for i in range(1,101):
    patients.append(
        Row(
            Patient_ID = f"P{i:04d}",
            DOB = date(1980 + (i % 30), (i % 12) + 1, (i % 28) + 1),
            Gender = genders[i % 2],
            City = cities[i % 5],
            Insurance_Type = insurance_types[i % 3],
            Registration_Date = date(2024, 1, 1) + timedelta(days=i * 3)
        )
    )

# Create PySpark DataFrame
patient_df = spark.createDataFrame(patients)

# Display the data
display(patient_df)
```

```python
patient_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("patients")
```

```python
patient_df.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/my_volume/patients_csv")
```

```python
from pyspark.sql import Row
from datetime import date, timedelta

encounters = []

encounter_types = ["OPD", "Emergency", "Inpatient"]
departments = ["Cardiology", "Orthopedics", "General Medicine", "Neurology"]

for i in range(1, 201):
    patient_number = ((i - 1) % 100) + 1
    
    encounters.append(
        Row(
            Encounter_ID=f"E{i:04d}",
            Patient_ID=f"P{patient_number:04d}",
            Encounter_Date=date(2024, 1, 1) + timedelta(days=i),
            Encounter_Type=encounter_types[i % 3],
            Department=departments[i % 4]
        )
    )

encounter_df = spark.createDataFrame(encounters)

display(encounter_df)
```

```python
encounter_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("encounters")
```

```python
encounter_df.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/my_volume/encounters_csv")
```

```python
from pyspark.sql import Row
from datetime import date, timedelta

claims = []

claim_statuses = ["Approved", "Pending", "Rejected"]
service_types = ["Consultation", "Diagnostic", "Procedure", "Pharmacy"]

for i in range(1, 301):
    patient_number = ((i - 1) % 100) + 1
    encounter_number = ((i - 1) % 200) + 1

    claims.append(
        Row(
            Claim_ID=f"C{i:04d}",
            Patient_ID=f"P{patient_number:04d}",
            Encounter_ID=f"E{encounter_number:04d}",
            Claim_Date=date(2024, 1, 1) + timedelta(days=i),
            Service_Type=service_types[i % 4],
            Claim_Amount=float(500 + (i % 20) * 250),
            Claim_Status=claim_statuses[i % 3]
        )
    )

claim_df = spark.createDataFrame(claims)

display(claim_df)
```

```python
claim_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("claims")
```

```python
claim_df.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/my_volume/claims_csv")
```

```python
from pyspark.sql import Row

providers = []

specialties = [
    "Cardiology",
    "Orthopedics",
    "General Medicine",
    "Neurology"
]

provider_cities = [
    "Hyderabad",
    "Bangalore",
    "Chennai",
    "Mumbai"
]

for i in range(1, 21):
    providers.append(
        Row(
            Provider_ID=f"PR{i:03d}",
            Provider_Name=f"Provider_{i:03d}",
            Specialty=specialties[(i - 1) % 4],
            City=provider_cities[(i - 1) % 4],
            Experience_Years=5 + (i % 16)
        )
    )

provider_df = spark.createDataFrame(providers)

display(provider_df)
```

```python
from pyspark.sql import Row
from datetime import date, timedelta

labs = []

test_names = [
    "Hemoglobin",
    "Blood Glucose",
    "Cholesterol",
    "Creatinine",
    "WBC"
]

for i in range(1, 401):
    patient_number = ((i - 1) % 100) + 1
    encounter_number = ((i - 1) % 200) + 1

    labs.append(
        Row(
            Lab_ID=f"L{i:04d}",
            Patient_ID=f"P{patient_number:04d}",
            Encounter_ID=f"E{encounter_number:04d}",
            Test_Date=date(2024, 1, 1) + timedelta(days=i),
            Test_Name=test_names[(i - 1) % 5],
            Test_Result=50 + (i % 51),
            Result_Status=["Normal", "High", "Low"][i % 3]
        )
    )

lab_df = spark.createDataFrame(labs)

display(lab_df)
```

```python
lab_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("labs")
```

```python
lab_df.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/my_volume/labs_csv")
```

```python
from pyspark.sql import Row
from datetime import date, timedelta

medications = []

medication_names = [
    "Paracetamol",
    "Amoxicillin",
    "Atorvastatin",
    "Metformin",
    "Amlodipine"
]

for i in range(1, 301):
    patient_number = ((i - 1) % 100) + 1
    encounter_number = ((i - 1) % 200) + 1

    medications.append(
        Row(
            Medication_ID=f"M{i:04d}",
            Patient_ID=f"P{patient_number:04d}",
            Encounter_ID=f"E{encounter_number:04d}",
            Medication_Date=date(2024, 1, 1) + timedelta(days=i),
            Medication_Name=medication_names[(i - 1) % 5],
            Dosage=f"{5 + (i % 6) * 5} mg",
            Duration_Days=5 + (i % 10)
        )
    )

medication_df = spark.createDataFrame(medications)

display(medication_df)
```

```python
medication_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("medications")
```

```python
medication_df.coalesce(1) \
    .write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/workspace/default/my_volume/medications_csv")
```
